```python
from anthropic import Anthropic
client = Anthropic()
model = "claude-sonnet-5-5"
```

```python
def add_user_message(messages, a_text):
    user_message = {"role": "user", "content": a_text}
    messages.append(user_message)

def add_assistant_message(messages, a_text):
    assistant_message = {"role": "assistant", "content": a_text}
    messages.append(assistant_message) 
```

```python
def chat(messages, system=None, stop_sequences=[]):
    params = {
        "model":model,
        "max_tokens":1000,
        # "thinking":{"type": "disabled"},
        "messages":messages
    }
    if system:
        params["system"] = system
    if stop_sequences:
        params["stop_sequences"] = stop_sequences

    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages = []

add_user_message(
    messages,
    "Generate 3 different sample aws cli commands. Each should be short. "
    "Output only the commands, one per line, with no markdown fences "
    "and no explanation. End your reply with the line END."
)
answer = chat(messages, stop_sequences=["END"])

print(answer)
```
