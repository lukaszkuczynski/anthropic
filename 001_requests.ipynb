```python
from anthropic import Anthropic
client = Anthropic()
model = "claude-haiku-5-5"
```

```python
def add_user_message(messages, a_text):
    user_message = {"role": "user", "content": a_text}
    messages.append(user_message)

def add_assistant_message(messages, a_text):
    assistant_message = {"role": "assistant", "content": a_text}
    messages.append(assistant_message) 
```

```python
def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=100,
        thinking={"type": "disabled"},
        messages=messages
    )
    print(message)
    return message.content[0].text
```

```python
messages = []

add_user_message(messages, "define quantum compute in one sentence")

answer = chat(messages)

add_assistant_message(messages, answer)

add_user_message(messages, "Write next sentence.")

answer = chat(messages)
answer
```

```python
messages = []

while True:
    user_input = input("> ")
    print(">", user_input)
    add_user_message(messages, user_input)
    assistant_message = chat(messages)
    add_assistant_message(messages, assistant_message)
    print(assistant_message)
```
