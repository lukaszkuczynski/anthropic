```python
from anthropic import Anthropic
client = Anthropic()
model = "claude-haiku-5-5"
```

```python
def add_user_message(messages, a_text):
    user_message = {"role": "user", "content": a_text}
    messages.append(user_message)

def add_assistant_message(messages, a_text):
    assistant_message = {"role": "assistant", "content": a_text}
    messages.append(assistant_message) 
```

```python
def chat(messages, system=None):
    params = {
        "model":model,
        "max_tokens":1000,
        "thinking":{"type": "disabled"},
        "messages":messages
    }
    if system:
        params["system"] = system
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages = []

add_user_message(messages, "Write a py functions that checks a string for duplicates")

system_concise = """
"Do not use comments, write as concise as possible"
"""

system_explain = """
Use comments, I am learning python and I have never used any coding language
"""

answer = chat(messages, system_explain)

print(answer)
```
